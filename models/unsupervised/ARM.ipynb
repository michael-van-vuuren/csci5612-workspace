```python
import time
import pandas as pd

def enrich_with_spotify_data(df):
    """Enhance DataFrame by adding 'Genre' and 'Followers' using Spotify API."""
    enriched_data = []
    
    i = 0
    total_rows = len(df)
    print('Processing rows for Spotify data:')
    
    for _, row in df.iterrows():
        isrc = row.get('ISRC')
        if pd.notna(isrc):
            result = spt.search_isrc(isrc, include_details=True)
            time.sleep(0.8)
            
            if result:
                row['Genre'] = ', '.join(result.get('Genres', []))
                row['Followers'] = result.get('Followers', 0)
            else:
                row['Genre'] = None
                row['Followers'] = None
        
        enriched_data.append(row.to_dict())
        i += 1
        print(f'{i}/{total_rows} --- {round((i/total_rows)*100)}% complete.')
    
    return pd.DataFrame(enriched_data)

df_enriched = enrich_with_spotify_data(df)
df_enriched.to_csv('../../data/arm-ready.csv', encoding='utf-8', index=False)
```

## Augment data

## Map subgenres to major genres

```python
genre_map = {
    'Pop': ['pop', 'dance pop', 'pop punk', 'pop rock', 'electropop', 'latin pop', 'k-pop', 'j-pop', 'bedroom pop', 'acoustic pop', 'indie pop', 'pop urbano', 'colombian pop', 'soft pop'],
    'Rock': ['rock', 'alternative rock', 'classic rock', 'garage rock', 'hard rock', 'metal', 'punk', 'emo', 'grunge', 'post-grunge', 'psychedelic rock', 'ska', 'ska punk', 'indie rock'],
    'Rap': ['rap', 'hip hop', 'trap', 'drill', 'grime', 'uk drill', 'west coast hip hop', 'east coast hip hop', 'underground hip hop', 'emo rap', 'rage rap', 'gangster rap', 'latin hip hop', 'experimental hip hop'],
    'R&B': ['r&b', 'neo soul', 'alternative r&b', 'motown', 'soul', 'latin r&b'],
    'Electronic': ['edm', 'electronic', 'house', 'techno', 'electro', 'dubstep', 'progressive house', 'melodic house', 'trance', 'synthwave', 'future house', 'eurodance'],
    'Country': ['country', 'country rock', 'country pop', 'acoustic country', 'bluegrass', 'red dirt'],
    'Reggaeton': ['reggaeton', 'reggaeton chileno', 'urbano latino', 'latin urban', 'neoperreo'],
    'Latin': ['latin', 'bachata', 'salsa', 'cumbia', 'ranchera', 'vallenato', 'flamenco', 'latin folk', 'tropical house']
}
def map_genre(genre_list):
    """Maps a subgenre to a major genre."""
    if not genre_list or genre_list == '[]':
        return 'Unknown'
    
    genre_list = genre_list.lower().split(', ')
    for broad_genre, subgenres in genre_map.items():
        if any(genre in subgenres for genre in genre_list):
            return broad_genre
    return 'Other'
```

## Convert record dataset to transactional dataset

```python
"""
Numerical features:
"""
num_cols = [
    'Spotify Streams',
    'TikTok Impact',
    'YouTube View-to-Like Ratio', 
    'TikTok View-to-Like Ratio', 
    'TikTok Likes per Post', 
    'Shazam Conversion Rate'
]

# bin numeric features into low & high
def bin_numeric(df, cols):
    for col in cols:
        df[col] = pd.qcut(df[col], q=2, labels=['Low', 'High'])
    return df
df = bin_numeric(df, num_cols)

"""
Categorical features (`Genre` and `Explicit Track`):
"""
# extract sub genres and explode them
df = df.rename(columns={'Genre': 'Sub Genre'})
df['Sub Genre'] = df['Sub Genre'].fillna('')
df['Sub Genre'] = df['Sub Genre'].str.split(', ')
df = df.explode('Sub Genre')

# apply mappings to reduce number of one-hot encoded features
df['Genre'] = df['Sub Genre'].apply(map_genre)
df = df[(df['Genre'] != 'Other') & (df['Genre'] != 'Unknown')]

df['Explicit Track'] = df['Explicit Track'].astype(str)

# mlxtend's apriori function requires transactional datasets to be one-hot encoded 
encoded_df = pd.get_dummies(df[['Explicit Track', 'Genre'] + num_cols])
encoded_df = encoded_df.rename(columns={'Explicit Track_0': 'Explicit_No', 'Explicit Track_1': 'Explicit_Yes'})
encoded_df.columns = encoded_df.columns.str.replace(' ', '_')
encoded_df = encoded_df.replace({0: False, 1: True})
encoded_df.head()
```

## Run Apriori for ARM

```python
# run apriori on the transactional dataset
frequent_itemsets = apriori(encoded_df, min_support=0.15, use_colnames=True)
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.5)

# extract top 15 rules for each metric
n = 15

top_support = rules.nlargest(n, 'support')
top_confidence = rules.nlargest(n, 'confidence')
top_lift = rules.nlargest(n, 'lift')

# display top 15 rules for each metric
print('Top 15 Rules by Support:')
display(top_support[['antecedents', 'consequents', 'support']])
print('\nTop 15 Rules by Confidence:')
display(top_confidence[['antecedents', 'consequents', 'confidence']])
print('\nTop 15 Rules by Lift:')
display(top_lift[['antecedents', 'consequents', 'lift']])
```

## Create graphs to visualize top rules

```python
def format_rule(rule):
    """Make rules easier to read."""
    parts = rule.rsplit('_', 1)
    return parts[0].replace('_', ' ') + ': ' + parts[1] if len(parts) > 1 else parts[0]

def draw_rule_network(network_metric_map, network_metric):
    """Draw network graph to visualize top rules based on support, confidence, and lift metrics."""
    
    # form directed rule network
    G = nx.DiGraph()

    for _, row in network_metric_map[network_metric].iterrows():
        antecedents = '\n'.join([format_rule(a) for a in sorted(row['antecedents'])])
        consequents = '\n'.join([format_rule(a) for a in sorted(row['consequents'])])
        G.add_edge(antecedents, consequents, weight=row[network_metric.lower()])

    # plot network
    plt.figure(figsize=(10, 8), dpi=300)
    pos = nx.spring_layout(G, iterations=30, k=0.64, seed=1111111)

    node_colors = []
    for node in G.nodes:
        has_incoming = G.in_degree(node) > 0
        has_outgoing = G.out_degree(node) > 0

        if has_incoming and has_outgoing:
            node_colors.append('#a1e6ceaa')  # both antecedent and consequent
        elif has_incoming:
            node_colors.append('#c1eba9aa')  # only consequent
        else:
            node_colors.append('#a1cce6aa')  # only antecedent

    nx.draw(G, pos, with_labels=True, node_color=node_colors, edge_color='lightgrey',
            node_size=1200, font_size=5, font_weight='bold', arrows=True, linewidths=0)

    edge_weights = nx.get_edge_attributes(G, 'weight')
    edge_labels = {k: f'{v:.2f}' for k, v in edge_weights.items()}
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, font_color='lightgrey', font_weight='bold')

    plt.title(f'Association Rules Network (Top {network_metric})')
    plt.show()

network_metrics = ['Support', 'Confidence', 'Lift']
network_metric_map = {'Support': top_support, 
                      'Confidence': top_confidence, 
                      'Lift': top_lift, 
                     }
for network_metric in network_metrics:
    draw_rule_network(network_metric_map, network_metric)
```
